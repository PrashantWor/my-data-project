### Task 2.1: Total Matches and Goals per League

```python
%%sql
--SELECT * FROM League
--LIMIT 5;
```

```python
%%sql
--SELECT * FROM Match
--LIMIT 5;
```

```python
%%sql 
SELECT 
 Country.id, 
 Country.name as country, 
 --League.id as league_id, 
 League.name as league_name, 
 sum(home_team_goal+away_team_goal) as total_goals_scored,
round(avg(home_team_goal+away_team_goal),1) as goals_per_match
FROM Country
INNER JOIN League
ON Country.id = League.country_id
INNER JOIN Match
ON Country.id = Match.country_id
GROUP BY Country.id, Country.name, League.id,League.name
ORDER By round(avg(home_team_goal+away_team_goal),1) DESC;
```

### Task 2.2: Home vs. Away Match Outcomes by Season

```python
%%sql
SELECT * FROM Country
Limit 10
```

```python
%%sql
SELECT 
 country_id, 
 Country.name AS Country, 
 season, 
 Match.home_team_api_id, 
 Match.away_team_api_id, 
 Team.team_long_name AS home_team, 
 Match.home_team_goal, 
 Match.away_team_goal,
CASE
 WHEN home_team_goal > away_team_goal THEN 'Home_won'
 WHEN home_team_goal < away_team_goal THEN 'Away_won'
 Else 'Draw'
END as Match_result
FROM Match
INNER JOIN Country
ON Match.country_id = Country.id
INNER JOIN Team
ON Match.home_team_api_id = Team.team_api_id
GROUP BY country_id, name, season,home_team_api_id, away_team_api_id,Team.team_long_name,home_team_goal, 
 away_team_goal;
```

### Task 2.3: Top 10 Highest-Scoring Home Teams

```python
%%sql
SELECT 
 Match.home_team_api_id, 
 Team.team_long_name, 
 Team.team_short_name,
 sum(Match.home_team_goal) as total_home_team_goals,
 sum(Match.away_team_goal) as total_away_team_goals
FROM Team
INNER JOIN Match
ON Team.team_api_id = Match.home_team_api_id
GROUP BY Match.home_team_api_id, Team.team_long_name, Team.team_short_name
HAVING count(home_team_api_id)>50
ORDER BY sum(Match.home_team_goal) DESC;
```

### Task 2.4: Tallest Players and Their Peak Overall Rating

```python
%%sql 
SELECT 
 Player.player_api_id, 
 Player.player_name, 
 ROUND(Player.height,1) as Player_height,
 birthday,
 strftime('%Y', 'now')- strftime('%Y', birthday) as Age,
 MAX(Player_Attributes.overall_rating) as player_overall_rating
FROM Player
INNER JOIN Player_Attributes
ON Player.player_api_id = Player_Attributes.player_api_id
WHERE height > 195
GROUP BY Player.player_api_id, player_name, height
ORDER BY Max(Player_Attributes.overall_rating) DESC
```
