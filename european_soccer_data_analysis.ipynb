```python
query = '''SELECT Player.id, Player.player_api_id, player_name, Country.name, birthday, overall_rating, potential, preferred_foot
FROM Player 
INNER JOIN Player_Attributes
ON  Player.player_api_id = Player_Attributes.player_api_id
INNER JOIN Country
ON Player.id=Country.id
'''
```

```python
result_df=pd.read_sql(query, conn)
display(result_df)
```
